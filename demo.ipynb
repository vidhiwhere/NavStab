# NavStab Interactive Demo

This Jupyter Notebook provides an executable demonstration of the **NavStab** hydrostatics and stability calculation engine.

NavStab is written in JavaScript/TypeScript. We'll use Python's `subprocess` module to run the examples directly within this notebook, giving you a smooth execution environment.

```python
import subprocess
import os

def run_navstab_script(script_name):
    """Utility function to execute a Node.js script and display its output."""
    print(f"Running {script_name}...\n")
    try:
        # Assuming Node.js is installed in the system PATH
        result = subprocess.run(['node', script_name], capture_output=True, text=True, check=True)
        print(result.stdout)
    except subprocess.CalledProcessError as e:
        print(f"❌ Error running {script_name}:")
        print(e.stderr)
    except FileNotFoundError:
        print("❌ Node.js is not installed or not in your PATH. Please install Node.js to run NavStab.")
```

## 1. Core Hydrostatics and Stability Calculations

The `example-usage.js` script demonstrates the core capabilities of the NavStab engine:
1. Defining ship parameters (LBP, B, T, etc.)
2. Generating a hull form and offset table
3. Calculating primary hydrostatic parameters (Displacement, LCB, KB, KM, etc.)
4. Generating a GZ (Righting Lever) curve for stability
5. Performing an IMO stability criteria check
6. Running a batch calculation across different drafts

Let's execute it now:

```python
run_navstab_script('example-usage.js')
```

## 2. Generating PDF Reports

NavStab can also generate detailed PDF reports, which is very useful for documentation. Let's run `generate-report-pdf.js` to create `sample-report.pdf`.

```python
run_navstab_script('generate-report-pdf.js')

if os.path.exists('sample-report.pdf'):
    print("✅ Success! 'sample-report.pdf' was generated in the current directory.")
else:
    print("⚠️ Report might not have been generated.")
```

## 3. Extending the Application
You can modify `example-usage.js` or create a new JavaScript file to calculate hydrostatics for your custom vessel and run it here using `run_navstab_script('your-script.js')`.
